##yf.download()

Es una función de la biblioteca yfinance en Python<br> que se utiliza para descargar archivos de datos financieros, <br>como precios de acciones, índices y otros datos financieros,<br> desde Yahoo Finance.Tambien permite descargar los datos en <br>formato CSV o pickle, lo que facilita su análisis y manipulación.

##SMA

sma_corto.png



La SMA (Media Móvil Simple) de 10 períodos es una <br>herramienta ágil que sigue cambios recientes en el precio, útil para<br> identificar señales rápidas en mercados volátiles. Por otro lado, la <br>SMA de 50 períodos proporciona una visión más estable y de largo<br> plazo al suavizar fluctuaciones menores, ayudando a confirmar <br>tendencias principales. Cuando la SMA corta cruza la larga al alza, <br>se interpreta como señal de compra; si cruza a la baja, como señal<br> de venta. Su uso conjunto ayuda a evaluar el impulso y las tendencias<br> del mercado, equilibrando rapidez y estabilidad.



##Stop Loss

stop-loss.png

Un Stop Loss (SL) es una orden de trading que te permite<br> sellar una posición abierta automáticamente cuando el precio<br> de un activo financieros alcanza un nivel determinado.<br> Su función es proteger tus ganancias y minimizar tus pérdidas.
